```python
import json
```

```python
def load_data(filename):
    with open(filename , "r") as f:
        data = json.load(f)
        return data
```

```python
data = load_data('data.json')
```

```python
def display_users(data):
    print("Users, their connections and their liked pages: ")
    for user in data['users']:
        print(f"ID: {user['id']} - {user['name']} is friends with, {user['friends']} and their liked pages {user['liked_pages']}")
    print("pages information:")
    for page in data['pages']:
        print(f"{page['id']}: {page['name']}")
```

```python
display_users(data)
```
