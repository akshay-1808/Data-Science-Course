```python
with open('finaldata.txt', encoding='utf-8') as f:
    data = f.read()
#data
```

```python
chunks = data.split("\n\n")
chunks = [c for c in chunks if len(c)>3]
```

```python
chunks[0]
```

```python
def parse_chunk(chunk):
    try:
        chunk = chunk.strip()
        sep_chunk = chunk.split('\n')
        username = sep_chunk[0]
        no_of_posts = int((sep_chunk[1].split(" post")[0]).replace(',',''))
        no_of_followers = float(sep_chunk[2].split(" follower")[0].replace(",","").replace("K","").replace("M",""))
        if("K" in sep_chunk[2]):
            no_of_followers = int(no_of_followers*1000)
        elif("M" in sep_chunk[2]):
            no_of_followers = int(no_of_followers*1000000)
        else:
            no_of_followers = int(no_of_followers)
            
        no_of_following = sep_chunk[3]

        no_of_following = float(sep_chunk[3].split(" following")[0].replace(",","").replace("K","").replace("M",""))
        if("K" in sep_chunk[3]):
            no_of_following = int(no_of_following*1000)
        elif("M" in sep_chunk[3]):
            no_of_following = int(no_of_following*1000000)
        else:
            no_of_following = int(no_of_following)

        
        name = sep_chunk[4]
        if(len(sep_chunk)>5):
            type_of_page = sep_chunk[5]
            bio = "\n".join(sep_chunk[6:])
        else:
            type_of_page = "Unknown"
            bio = ''
        # print(username, no_of_posts, no_of_followers, no_of_following, name, type_of_page, bio, sep = '\n')
        return {"username": username, "no_of_posts": no_of_posts, "no_of_followers": no_of_followers, "no_of_following": no_of_following, "name": name,
                "type_of_page": type_of_page, "bio": bio}
    except Exception as e:
        print(e, chunk)

parse_chunk(chunks[0])
```

```python
all_chunks = []

for chunk in chunks:
    #print(f"The chunks is {chunk}")
    parsed_chunk = parse_chunk(chunk)
    all_chunks.append(parsed_chunk)

# print(all_chunks)
```

```python
import json
with open('final_data.json', 'w') as init_data:
    json.dump(all_chunks, init_data, indent = 4)
s = json.dumps(all_chunks, indent = 4)
# print(s)
```

```python
max_posts = 0
max_post_chunk = {}
max_followers = 0
max_follower_chunk = {}
max_following = 0
max_following_chunk = {}
categories = []
for chunk in all_chunks:
    if(max_posts< int(chunk['no_of_posts'])):
        max_posts = int(chunk['no_of_posts'])
        max_post_chunk = chunk
    if(max_followers< int(chunk['no_of_followers'])):
        max_followers = int(chunk['no_of_followers'])
        max_follower_chunk = chunk
    if(max_following< int(chunk['no_of_following'])):
        max_following = int(chunk['no_of_following'])
        max_following_chunk = chunk
    if((chunk['type_of_page'] == "Unknown")):
        continue
    else:
        if chunk['type_of_page'] not in categories:
            categories.append(chunk['type_of_page'])
    
print(f"Username: {max_post_chunk['username']} \nName: {max_post_chunk['name']} \nhas Maximum posts: {max_posts}\n")
print(f"Username: {max_follower_chunk['username']} \nName: {max_follower_chunk['name']} \nhas Maximum Followers: {max_followers}\n")
print(f"Username: {max_following_chunk['username']} \nName: {max_following_chunk['name']} \nis following Maximum people: {max_following}\n")
print("No. of categories: ", len(categories))
print(f"Categories: {categories}")
```
